## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
import pandas as pd

url = "https://raw.githubusercontent.com/MohimaAliKhan/FlyRank-MLInternship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Compute 3 key descriptive stats to demonstrate content variance across archetypes
total_pages = len(df)
high_impression_pages = len(df[df['impressions_90d'] > 10000])
avg_ctr = df['ctr'].mean()
decaying_pages_pct = (df['trend_direction'] == 'down').mean() * 100

print(f"Total Pages Analyzed: {total_pages:,}")
print(f"High Impression Pages (>10k 90d impressions): {high_impression_pages:,} ({high_impression_pages/total_pages*100:.1f}%)")
print(f"Average CTR across inventory: {avg_ctr:.2%}")
print(f"Percentage of pages with declining trend: {decaying_pages_pct:.1f}%")
```
